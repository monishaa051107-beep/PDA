```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv(r"C:\Users\DELL\Desktop\PDA\laptop_price_dataset_v4.csv")

# Correlation analysis
plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.title("Correlation Matrix")
plt.show()
```

```python
sns.countplot(x='category', data=df)

plt.title("Laptop Categories")
plt.show()
```

```python
sns.boxplot(x='graphics_card', y='price', data=df)

plt.xticks(rotation=45)
plt.title("GPU vs Price")
plt.show()
```

```python
sns.scatterplot(x='size_inch', y='price', data=df)

plt.title("Screen Size vs Price")
plt.show()
```

```python
sns.scatterplot(x='weight_kg', y='battery_wh', data=df)

plt.title("Weight vs Battery Capacity")
plt.show()
```

```python
df['price_per_ram'] = df['price'] / df['ram']
```

```python
sns.histplot(df['price_per_ram'])
plt.title("Price per RAM Distribution")
plt.show()
```

```python
df.groupby('brand')['price'].mean().sort_values(ascending=False)
```

```python
sns.barplot(x='brand', y='price', data=df)
plt.xticks(rotation=45)
plt.title("Average Laptop Price by Brand")
plt.show()
```

```python
df.groupby('category')['price'].mean()
```

```python
sns.boxplot(x='category', y='price', data=df)

plt.title("Laptop Price by Category")
plt.show()
```

```python
sns.scatterplot(x='ram', y='price', data=df)

plt.title("RAM vs Laptop Price")
plt.show()
```

```python
df[['ram','price']].corr()
```

```python
sns.boxplot(x='graphics_card', y='price', data=df)

plt.xticks(rotation=45)
plt.title("GPU Type vs Laptop Price")
plt.show()
```

```python
df.groupby('cpu')['price'].mean().sort_values(ascending=False)
```

```python
sns.barplot(x='cpu', y='price', data=df)

plt.xticks(rotation=45)
plt.title("Average Price by CPU Type")
plt.show()
```
